task1:
Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.


```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [5, 7, 6, 8, 10, 12, 9]
swiggy = [4, 5, 7, 6, 8, 9, 11]

plt.bar(days, zomato, color="orange", label="Zomato")
plt.bar(days, swiggy, bottom=zomato, color="purple", label="Swiggy")

plt.title("Daily Food Orders")
plt.xlabel("Days")
plt.ylabel("Total Orders")
plt.legend()

plt.show()
```

task2:
Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using Matplotlib's fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [100, 120, 110, 140, 160, 180, 170]
youtube = [200, 220, 210, 250, 270, 300, 290]

plt.figure(figsize=(8, 5))

plt.fill_between(days, spotify, color="green", alpha=0.5, label="Spotify")
plt.fill_between(days, youtube, color="red", alpha=0.4, label="YouTube")

plt.title("Daily Active Users")
plt.xlabel("Days")
plt.ylabel("Active Users")
plt.legend()

plt.show()
```

task3:
Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>


```python
days = list(range(1, 11))

delivery_time = [30, 35, 32, 40, 38, 33, 36, 42, 37, 34]
std_dev = [3, 4, 2, 5, 3, 4, 2, 6, 3, 4]

plt.figure(figsize=(8, 5))

plt.errorbar(
    days,
    delivery_time,
    yerr=std_dev,
    fmt="o-",
    color="blue",
    capsize=5,
    label="Delivery Time"
)

plt.title("Flipkart Average Delivery Time")
plt.xlabel("Day")
plt.ylabel("Delivery Time (Minutes)")
plt.legend()
plt.grid(True)

plt.show()
```

task4:
Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').


```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta

today = datetime.today().date()
dates = [today - timedelta(days=i) for i in range(13, -1, -1)]

balance = [2500, 2300, 2800, 2600, 3000, 2700, 3200,
           2900, 3500, 3300, 3800, 3600, 4000, 4200]

plt.figure(figsize=(10, 5))

plt.plot(dates, balance, marker="o", color="green")

plt.title("Paytm Wallet Balance - Last 14 Days")
plt.xlabel("Date")
plt.ylabel("Wallet Balance (₹)")

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))
plt.gcf().autofmt_xdate()

plt.grid(True)
plt.show()
```

task5:
Refactor your area chart code to add gridlines and custom axis titles, making the chart look more like the analytics dashboards you see in apps like Instagram Insights.<br><br><em><strong>Constraint:</strong> Use ax.set_title(), ax.set_xlabel(), ax.set_ylabel(), and ax.grid(True) in your code.</em>


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

instagram = [100, 130, 120, 150, 180, 200, 190]
youtube = [200, 220, 250, 230, 280, 300, 320]

fig, ax = plt.subplots(figsize=(9, 5))

ax.fill_between(days, instagram, color="magenta",
                alpha=0.4, label="Instagram")
ax.fill_between(days, youtube, color="cyan",
                alpha=0.3, label="YouTube")

ax.set_title("Social Media Analytics Dashboard")
ax.set_xlabel("Days of the Week")
ax.set_ylabel("Daily Active Users")

ax.grid(True)
ax.legend()

plt.tight_layout()
plt.show()
```
