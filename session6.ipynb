task2:
Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>


```python
friends = ["Amit", "Riya", "Jay", "Neha", "Raj", "Priya",
           "Meet", "Karan", "Pooja", "Dev", "Ravi", "Mira"]

spending = [500, 800, 1200, 600, 1500, 900, 700, 1800, 1100, 400, 1300, 1000]
orders = [2, 3, 5, 2, 6, 4, 3, 7, 4, 1, 5, 4]

plt.scatter(orders, spending)

for i in range(len(friends)):
    plt.annotate(friends[i], (orders[i], spending[i]))

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Spending vs Orders")

plt.show()
```

task3:
Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>


```python
spending = [500, 800, 1200, 600, 1500, 900, 700, 1800, 1100, 400, 1300, 1000]
orders = [2, 3, 5, 2, 6, 4, 3, 7, 4, 1, 5, 4]

slope, intercept = np.polyfit(orders, spending, 1)

predicted = np.polyval([slope, intercept], orders)

plt.scatter(orders, spending, label="Friends' Data")
plt.plot(orders, predicted, color="red", label="Trend Line")

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Orders with Trend Line")
plt.legend()
plt.show()

print("Slope:", slope)
print("Intercept:", intercept)
```

task4:
Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.


```python
youtube = [1, 2, 3, 4, 2, 5, 6, 3, 4, 7, 2, 5, 6, 1, 4]
spotify = [2, 3, 4, 5, 3, 6, 7, 4, 5, 8, 3, 6, 7, 2, 5]

slope, intercept = np.polyfit(youtube, spotify, 1)

predicted = np.polyval([slope, intercept], youtube)

plt.scatter(youtube, spotify, label="Users")
plt.plot(youtube, predicted, color="red", label="Regression Line")

equation = f"y = {slope:.2f}x + {intercept:.2f}"
plt.text(1, 7.5, equation, fontsize=10)

plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.title("YouTube vs Spotify Usage")
plt.legend()
plt.show()

print("Regression Equation:", equation)
```
