task1 : 
Use plt.hist() to plot a histogram of delivery times (in minutes) for 30 Swiggy orders:
 [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 35, 40, 38, 33, 30, 42].
  Show the plot window.


```python
import matplotlib.pyplot as plt

delivery_times = [
    32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
    27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
    36, 28, 43, 31, 35, 40, 38, 33, 30, 42
]

plt.hist(delivery_times, bins=5, color="orange", edgecolor="black")

plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")
plt.title("Swiggy Delivery Time Distribution")

plt.show()
```

task2 :
Plot a histogram of IPL match scores (runs per team in a match) using plt.hist() and experiment with bins=5, bins=10, and bins=15. Observe and describe how changing the number of bins affects the shape of the histogram.<br><br><em><strong>Hint:</strong> Use any sample array of 50-60 scores between 120 and 220.</em>


```python
scores = [
    145, 167, 182, 155, 198, 176, 134, 210, 189, 165,
    172, 158, 201, 143, 187, 194, 160, 178, 220, 151,
    169, 185, 132, 205, 174, 191, 148, 163, 199, 180,
    156, 212, 138, 170, 186, 196, 147, 162, 208, 177,
    153, 184, 193, 141, 166, 202, 159, 175, 188, 130
]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.hist(scores, bins=5, edgecolor="black")
plt.title("Bins = 5")

plt.subplot(1, 3, 2)
plt.hist(scores, bins=10, edgecolor="black")
plt.title("Bins = 10")

plt.subplot(1, 3, 3)
plt.hist(scores, bins=15, edgecolor="black")
plt.title("Bins = 15")

plt.tight_layout()
plt.show()
```

task3:
Given an array of daily step counts from a fitness tracker app for one month, plot two histograms using plt.hist(): one showing frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.


```python
steps = [
    4000, 5000, 6000, 4500, 7000, 8000, 5500,
    6500, 7500, 5000, 6000, 8500, 9000, 4500,
    7000, 6500, 5500, 8000, 7500, 6000, 5000,
    9500, 7000, 6500, 8500, 5500, 9000, 7500,
    6000, 8000
]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(steps, bins=5, edgecolor="black")
plt.xlabel("Daily Steps")
plt.ylabel("Frequency")
plt.title("Frequency Histogram")

plt.subplot(1, 2, 2)
plt.hist(steps, bins=5, density=True, edgecolor="black")
plt.xlabel("Daily Steps")
plt.ylabel("Density")
plt.title("Density Histogram")

plt.tight_layout()
plt.show()
```

task4:
Simulate the distribution of song durations (in seconds) from your Spotify playlist by generating 50 random numbers between 120 and 300. Plot a histogram using plt.hist() and set the color to 'purple' and add the title 'Spotify Song Duration Distribution'.


```python
import random

random.seed(42)

durations = [random.randint(120, 300) for i in range(50)]

plt.hist(durations, bins=10, color="purple", edgecolor="black")

plt.xlabel("Song Duration (Seconds)")
plt.ylabel("Number of Songs")
plt.title("Spotify Song Duration Distribution")

plt.show()
```
