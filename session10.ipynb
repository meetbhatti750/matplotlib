task1:
Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

users = [1200, 1500, 1400, 1800, 2000, 2200,
         2100, 2500, 2300, 2800, 2600, 3200]

plt.figure(figsize=(10, 5))
plt.plot(months, users, marker="o", color="green")

max_index = users.index(max(users))

plt.annotate(
    "Highest Users",
    xy=(months[max_index], users[max_index]),
    xytext=(-70, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", color="red"),
    color="red"
)

plt.title("Monthly Active Users - Music App")
plt.xlabel("Months")
plt.ylabel("Active Users")
plt.grid(True)
plt.show()
```

task2:
Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>


```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

orders = [500, 450, 600, 700, 550, 800,
          750, 900, 650, 1000, 850, 950]

plt.figure(figsize=(10, 5))
plt.bar(months, orders, color="orange")

min_index = orders.index(min(orders))

plt.annotate(
    "Lowest Orders",
    xy=(min_index, orders[min_index]),
    xytext=(min_index + 1, orders[min_index] + 150),
    arrowprops=dict(arrowstyle="->", color="red"),
    color="red"
)

plt.title("Monthly Food Orders - Zomato")
plt.xlabel("Months")
plt.ylabel("Number of Orders")
plt.show()
```

task3:
Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.


```python
methods = ["UPI", "Card", "Wallet", "Cash"]
percentage = [50, 25, 15, 10]

plt.figure(figsize=(8, 6))

plt.pie(
    percentage,
    labels=methods,
    autopct="%1.1f%%",
    startangle=90
)

plt.legend(
    methods,
    loc="lower right",
    title="Payment Methods"
)

plt.title("Payment Methods - Flipkart")
plt.show()
```

task4:
Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>


```python
import matplotlib.pyplot as plt

movies = ["Movie A", "Movie B", "Movie C",
          "Movie D", "Movie E"]

ratings = [7.5, 8.2, 9.1, 6.8, 8.7]
reviews = [1200, 2500, 1800, 900, 3000]

plt.figure(figsize=(9, 5))
plt.scatter(ratings, reviews, color="purple", s=100)

max_index = ratings.index(max(ratings))

plt.annotate(
    movies[max_index] + " - Highest Rating",
    xy=(ratings[max_index], reviews[max_index]),
    xytext=(6.7, 3200),
    arrowprops=dict(arrowstyle="->", color="red"),
    bbox=dict(boxstyle="round,pad=0.5",
              facecolor="yellow", edgecolor="black")
)

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.grid(True)
plt.show()
```
