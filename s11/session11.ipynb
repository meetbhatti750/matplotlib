task1:
Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.


```python
import matplotlib.pyplot as plt

restaurants = ["Restaurant A", "Restaurant B", "Restaurant C",
               "Restaurant D", "Restaurant E"]
orders = [25, 40, 30, 50, 35]

plt.figure(figsize=(8, 5))
plt.bar(restaurants, orders, color="orange")

plt.title("Zomato Restaurant Orders")
plt.xlabel("Restaurants")
plt.ylabel("Number of Orders")
plt.xticks(rotation=20)

plt.tight_layout()
plt.savefig("zomato_orders.png", dpi=150)
plt.show()
```

task2:
Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>


```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5000, 6500, 7000, 6000, 8500, 9500, 10000]

plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker="o", color="blue")

plt.title("Daily Step Count")
plt.xlabel("Days")
plt.ylabel("Number of Steps")

plt.tight_layout()
plt.savefig("daily_steps.svg", format="svg")
plt.show()
```

task3:
Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().


```python
import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
hours = [3, 4, 2, 1]

plt.figure(figsize=(7, 6))

plt.pie(hours, labels=apps, autopct="%1.1f%%",
        startangle=90)

plt.title("Daily App Usage")
plt.savefig("app_usage.pdf", format="pdf",
            bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pptx import Presentation
from pptx.util import Inches

zometo_orders = sns.load_dataset("titanic")

print(zometo_orders.head())

plt.figure(figsize=(7, 5))

sns.boxplot(data=zometo_orders, x="pclass", y="fare")
plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

chart_image_path = "my_chart.png"
plt.savefig(chart_image_path, dpi=1200, bbox_inches="tight")
plt.show()
plt.close()

prs = Presentation()
slide = prs.slides.add_slide(prs.slide_layouts[6])

slide.shapes.add_picture(
    chart_image_path,
    Inches(1),
    Inches(1),
    width=Inches(8)
)

pptx_output_path = "my_chart.pptx"
prs.save(pptx_output_path)

print("Chart saved as", chart_image_path)
print("PowerPoint saved as", pptx_output_path)
```

task5:
Export a Matplotlib chart as a PDF, then verify the PDF opens correctly and the chart looks sharp (not blurry) at 200% zoom. If the image looks pixelated, try exporting again with a higher DPI setting and compare the results.


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.savefig("sales_chart.pdf", format="pdf",
            dpi=300, bbox_inches="tight")

plt.show()

print("PDF saved successfully!")
```
